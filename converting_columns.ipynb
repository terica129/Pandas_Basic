### Converting Columns

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_json('data/cars.json')
```

```python
df.head()
```

### Convert Pandas Column to list using .tolist()

```python
col_list = df['Car'].values.tolist()
print('CarLists:', col_list)
```

```python
df.Car.values.tolist()
```

### Convert Pandas column to a Numpy Array

```python
col_numpy_arr = df['Car'].to_numpy()
print(type(col_numpy_arr))
```

```python
df[df.columns[0]].values.tolist()
```

```python
# List of df indexes
df.index.tolist()
```

### Rename a column in Pandas

- .rename()

```python
tech = {
    'Courses': ["Python", ' DataAnalytics', 'MERN Stack', 'Data Engineering'],
    'Fee': [100, 2000, 250, 300],
    'Duration':['40days', '50days', '3 months', '1 year']
}

tech_df = pd.DataFrame(tech)
```

```python
tech_df = df.rename(columns = {'Courses': 'Course Name'})
```
